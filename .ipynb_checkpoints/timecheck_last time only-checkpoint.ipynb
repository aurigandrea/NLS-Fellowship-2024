```python
import requests
import pandas as pd
```

```python
# === CONFIG ===
input_csv_path = "SOI_music only.csv"  # Your input file with a column named 'URL'
output_csv_path = "SOI_music_urls_with_last_live.csv"  # Output file
```

```python
# === FUNCTION TO QUERY WAYBACK MACHINE ===
def get_last_live_capture(url):
    base = "http://web.archive.org/cdx/search/cdx"
    params = {
        'url': url,
        'output': 'json',
        'filter': 'statuscode:200',
        'limit': 1,
        'collapse': 'digest',
        'sort': 'descending'
    }
    try:
        response = requests.get(base, params=params, timeout=10)
        data = response.json()
        if len(data) > 1:
            timestamp = data[1][1]
            return f"{timestamp[:4]}-{timestamp[4:6]}-{timestamp[6:8]}"
        else:
            return "No archived live version"
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# === LOAD INPUT CSV ===
df = pd.read_csv(input_csv_path)

# === APPLY FUNCTION TO URL COLUMN ===
df['Last_Live_Capture'] = df['url'].apply(get_last_live_capture)
```

```python
# === SAVE OUTPUT CSV ===
df.to_csv(output_csv_path, index=False)
print(f"Saved updated CSV to {output_csv_path}")
```
