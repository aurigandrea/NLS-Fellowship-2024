```python
import pandas as pd
import requests
from tqdm import tqdm
import os
import time

```

```python
# === CONFIG ===
INPUT_CSV = "music.csv"  # Your original CSV with a 'URL' column
OUTPUT_CSV = "music_with_first_and_last_live.csv"  # Output with progress
SAVE_EVERY = 50  # Save progress every N URLs
SLEEP_BETWEEN = 1  # Seconds to wait between requests (avoid rate limits)

```

```python

# === FUNCTION TO QUERY WAYBACK MACHINE ===
def get_live_capture(url, sort_order='ascending'):
    base = "http://web.archive.org/cdx/search/cdx"
    params = {
        'url': url,
        'output': 'json',
        'filter': 'statuscode:200',
        'limit': 1,
        'collapse': 'digest',
        'sort': sort_order
    }
    try:
        response = requests.get(base, params=params, timeout=10)
        data = response.json()
        if len(data) > 1:
            ts = data[1][1]
            return f"{ts[:4]}-{ts[4:6]}-{ts[6:8]}"
        else:
            return "No archived live version"
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
# === LOAD DATA ===
df = pd.read_csv(INPUT_CSV)
df['First_Live_Capture'] = None
df['Last_Live_Capture'] = None
```

```python
# === PROCESS WITH PROGRESS BAR ===
print(f"🔍 Processing {len(df)} URLs...\n")
for i in tqdm(range(len(df)), desc="Processing URLs"):
    url = df.at[i, 'url']
    df.at[i, 'First_Live_Capture'] = get_live_capture(url, 'ascending')
    time.sleep(SLEEP_BETWEEN)
    df.at[i, 'Last_Live_Capture'] = get_live_capture(url, 'descending')
    time.sleep(SLEEP_BETWEEN)

    # Auto-save every N rows
    if i % SAVE_EVERY == 0 and i > 0:
        df.to_csv(OUTPUT_CSV, index=False)

# === FINAL SAVE ===
df.to_csv(OUTPUT_CSV, index=False)
print(f"\n✅ Done! Full output saved to: {OUTPUT_CSV}")
```
