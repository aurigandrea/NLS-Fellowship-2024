```python
import pandas as pd
import requests
from tqdm import tqdm

# === CONFIG ===
input_csv_path = "SOI_music_all.csv"  # Input CSV with a column named 'URL'
output_csv_path = "SOI_music_with_first_and_last_live.csv"  # Output file

# === FUNCTION TO QUERY WAYBACK MACHINE FOR FIRST OR LAST CAPTURE ===
def get_live_capture(url, sort_order='ascending'):
    base = "http://web.archive.org/cdx/search/cdx"
    params = {
        'url': url,
        'output': 'json',
        'filter': 'statuscode:200',
        'limit': 1,
        'collapse': 'digest',
        'sort': sort_order
    }
    try:
        response = requests.get(base, params=params, timeout=10)
        data = response.json()
        if len(data) > 1:
            timestamp = data[1][1]
            return f"{timestamp[:4]}-{timestamp[4:6]}-{timestamp[6:8]}"
        else:
            return "No archived live version"
    except Exception as e:
        return f"Error: {str(e)}"

```

```python
# === LOAD CSV ===
df = pd.read_csv(input_csv_path)

# === INITIALISE PROGRESS BAR ===
tqdm.pandas(desc="Checking Wayback Machine")


```

```python

# === APPLY WITH PROGRESS BAR ===
df['First_Live_Capture'] = df['url'].progress_apply(lambda url: get_live_capture(url, 'ascending'))
df['Last_Live_Capture'] = df['url'].progress_apply(lambda url: get_live_capture(url, 'descending'))

# === SAVE OUTPUT ===
df.to_csv(output_csv_path, index=False)
print(f"\n✅ Saved updated CSV to {output_csv_path}")
```
